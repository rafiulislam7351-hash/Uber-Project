## Saving Silver Data

```python
# 1. Read your DLT silver table
df = spark.read.table("uberdata.silver.transition_stream")

# 2. Define your export path
target_path = "abfss://silver@rideinformation.dfs.core.windows.net/transition_stream"

# 3. Write directly to ADLS using Delta or Parquet format
df.write \
  .format("delta") \
  .mode("append") \
  .save(target_path)
```

## Saving Gold as a Table

```python
# 1. Read your DLT gold table
df = spark.read.table("uberdata.gold.dim_booking")

# 2. Define your export path and gold table details
target_path = "abfss://gold@rideinformation.dfs.core.windows.net/dim_booking"
gold_table_name = "uberdata.gold.dim_booking"

# 3. Create table if it doesn't exist (or append if it exists)
df.write \
  .format("delta") \
  .mode("append") \
  .option("path", target_path) \
  .save()
```
